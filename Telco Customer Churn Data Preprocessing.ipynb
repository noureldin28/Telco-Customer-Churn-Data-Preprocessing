## Part 4: Outlier Detection & Handling

1.  **Identify Numerical Columns:**

```python
numerical_columns = df_filtered.select_dtypes('number').columns.tolist()
numerical_columns
```

2.  **Visualize `MonthlyCharges` (Boxplot):**

```python
sns.boxplot(x='MonthlyCharges',data=df_filtered,orient='h')
plt.show()
```

3.  **Handle `MonthlyCharges` Outliers (IQR Method):**

```python
#MonthlyCarges has no outliers but if it had, that would be the code to handle them.
Q1_mon_charges = df_filtered['MonthlyCharges'].quantile(0.25)
Q3_mon_charges = df_filtered['MonthlyCharges'].quantile(0.75)
IQR_mon_charges = Q3_mon_charges - Q1_mon_charges

upper_bound_mon_charges = Q3_mon_charges + 1.5 * IQR_mon_charges
lower_bound_mon_charges = Q1_mon_charges - 1.5 * IQR_mon_charges

df_filtered['MonthlyCharges'] = df_filtered['MonthlyCharges'].clip(lower=lower_bound_mon_charges,upper=upper_bound_mon_charges)
```

4.  **Visualize `TotalCharges` (Boxplot):**

```python
sns.boxplot(x="TotalCharges", data=df_filtered, orient="h")
plt.show()
```

5.  **Handle `TotalCharges` Outliers (IQR Method):**

```python
#TotalCharges has no outliers but if it had, that would be the code to handle them.
Q1_total_charges = df_filtered['TotalCharges'].quantile(0.25)
Q3_total_charges = df_filtered['TotalCharges'].quantile(0.75)
IQR_total_charges = Q3_total_charges - Q1_total_charges

upper_bound_total_charges = Q3_total_charges + 1.5 * IQR_total_charges
lower_bound_total_charges = Q1_total_charges - 1.5 * IQR_total_charges

df_filtered['TotalCharges'] = df_filtered['TotalCharges'].clip(lower=lower_bound_total_charges,upper=upper_bound_total_charges)
```

6.  **Re-visualize `TotalCharges`:**

```python
sns.boxplot(x="TotalCharges", data=df_filtered, orient="h")
plt.show()
```

```python
columns_to_encode = ['InternetService','Contract','PaymentMethod',
                     'OnlineSecurity','OnlineBackup','DeviceProtection',
                     'TechSupport','StreamingTV','StreamingMovies']
encoded_df = pd.get_dummies(
    data=df_filtered,
    columns = columns_to_encode,
    dtype=int
)

#binary columns have only yes or no values that is why i mapped them.
binary_columns = ['Partner','Dependents','PaperlessBilling','Churn']
for i in binary_columns:
    encoded_df[i] = encoded_df[i].map({
        'Yes':1,
        'No':0
        }).astype(int)

encoded_df.head()
```

**Encoding Categorical Variables:** I chose to encode the categorical variables because machine-learning algorithms generally require numerical input. One-hot encoding was used for categorical variables with multiple categories, while binary variables containing only “Yes” and “No” were mapped to 1 and 0. This creates a numerical dataset that can be used for future machine-learning modeling.
